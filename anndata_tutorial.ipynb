```python
# This is a anndata tutorial Python script.
# Press ⌃R to execute it or replace it with your code.

import numpy as np
import pandas as pd
import anndata as ad
import scanpy as sc
from scipy.sparse import csr_matrix
from importlib.metadata import version
from IPython.display import display
version("anndata")
```

```python
def print_hi(name):
    print(f'Hi, {name}')  # Press ⌘F8 to toggle the breakpoint.

if __name__ == '__main__':
    print_hi('PyCharm')
```

```python
counts = csr_matrix(np.random.poisson(1, size=(100, 2000)), dtype=np.float32)
adata = ad.AnnData(counts)
print("adata: ", adata)
print("adata.X: ", adata.X)
```

```python
adata.obs_names = [f"Cell_{i:d}" for i in range(adata.n_obs)]
adata.var_names = [f"Gene_{i:d}" for i in range(adata.n_vars)]
print(adata.obs_names[:10])
print(adata.var_names[:10])

print("subset:" , adata[["Cell_1", "Cell_10"], ["Gene_1", "Gene_20", "Gene_1981"]])
```

```python
ct = np.random.choice(["B", "T", "Monocyte"], size = (adata.n_obs))
adata.obs["Cell_type"] = pd.Categorical(ct)
display(adata.obs)
```

```python
display(adata)
print(adata)
```

```python
print("n_obs: ", adata.n_obs)
print("n_vars: ", adata.n_vars)
gene_types = np.random.choice(["intron", "exon", "promotor"], size=adata.n_vars)
adata.var["gene_type"] = pd.Categorical(gene_types)
display(adata)
display(adata.var)
```

```python
adata.obsm["X_umap"] = np.random.normal(0, 1, size=(adata.n_obs, 2))
adata.varm["gene_stuff"] = np.random.normal(0, 1, size=(adata.n_vars, 5))

print(adata)

print(adata.obsm["X_umap"][:5])
print(adata.varm["gene_stuff"][:5])
```

```python
adata.uns["random"] = [1, 2, 3]
adata.uns
```

```python
adata.layers["log_transformed"] = np.log1p(adata.X)
adata
```

```python
adata.to_df(layer="log_transformed")
```

```python
# adata.write_zarr("my_results.zarr")
adata.write_h5ad("my_results.h5ad")
```

```python
adata_file = sc.read_h5ad("my_results.h5ad")
print(adata_file)

print("Shape:", adata_file.shape)
print("Cells:", adata_file.n_obs)
print("Genes:", adata_file.n_vars)
```
